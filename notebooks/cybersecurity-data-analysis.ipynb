### Outlier Analysis

```python
Q1 = df_attack['Total Fwd Packets'].quantile(0.25)
Q3 = df_attack['Total Fwd Packets'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outlier_mask = (
    (df_attack['Total Fwd Packets'] < lower_bound) |
    (df_attack['Total Fwd Packets'] > upper_bound)
)

outlier_count = outlier_mask.sum()

print("Jumlah outlier:", outlier_count)
print("Persentase:", round(outlier_count / len(df_attack) * 100, 2), "%")
```

```python
df_attack[outlier_mask]['Label'].value_counts()
```

```python
df_attack[outlier_mask].groupby('Label')['Total Fwd Packets'].agg(
    ['count', 'mean', 'median', 'max']
)
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_attack,
    x='Label',
    y='Total Fwd Packets'
)

plt.title('Outlier Analysis: Total Forward Packets')
plt.xlabel('Attack Type')
plt.ylabel('Total Forward Packets')
plt.xticks(rotation=45)

plt.show()
```

Finding: Sebanyak 839 records (3.75%) teridentifikasi sebagai outlier berdasarkan Total Fwd Packets menggunakan metode IQR. Mayoritas outlier berasal dari BruteForce dengan 582 records.

Statistical Insight: Infiltration memiliki perbedaan yang sangat besar antara mean (1,029.66) dan median (35), menunjukkan bahwa beberapa nilai ekstrem sangat memengaruhi nilai rata-ratanya.

Security Insight: Flow dengan jumlah forward packet yang sangat tinggi dapat menjadi kandidat untuk investigasi lebih lanjut karena menunjukkan pola traffic yang berbeda dari mayoritas flow. Namun, outlier secara statistik tidak otomatis berarti malicious atau lebih berbahaya.
